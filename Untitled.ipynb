```python
%pip install mygene
%pip install mygene httpx

import mygene
mg = mygene.MyGeneInfo()
def combine_values(values):
    """Join unique non-empty semicolon-separated values."""
    return "; ".join(sorted({
        item.strip()
        for value in values
        if pd.notna(value)
        for item in str(value).split(";")
        if item.strip()
    }))


def extract_ensembl_ids(value):
    """Return Ensembl gene IDs from a MyGene result."""
    records = value if isinstance(value, list) else [value]
    return "; ".join(sorted({
        str(record["gene"])
        for record in records
        if isinstance(record, dict) and record.get("gene")
    }))


def clean_mim_number(value):
    """Convert Excel-style MIM values such as 607872.0 to 607872."""
    if pd.isna(value):
        return ""
    return re.sub(r"\.0$", "", str(value).strip())


def map_to_ensembl(values, scope, input_column, symbol_column, cleaner=str):
    """Map unique human gene symbols or MIM IDs through MyGene.info."""
    values = sorted({
        cleaned
        for value in values
        if (cleaned := cleaner(value).strip())
    })

    columns = [input_column, symbol_column, "Entrez_ID", "Ensembl_ID"]

    print(f"  Mapping {len(values):,} unique {scope} values...")

    if not values:
        return pd.DataFrame(columns=columns)

    results = mg.querymany(
        values,
        scopes=scope,
        fields="symbol,entrezgene,ensembl.gene",
        species="human",
        verbose=False,
    )

    rows = [
        {
            input_column: cleaner(result.get("query", "")),
            symbol_column: str(result.get("symbol") or ""),
            "Entrez_ID": str(result.get("entrezgene") or ""),
            "Ensembl_ID": extract_ensembl_ids(result.get("ensembl")),
        }
        for result in results
    ]

    return (
        pd.DataFrame(rows, columns=columns)
        .groupby(input_column, as_index=False)
        .agg({column: combine_values for column in columns[1:]})
    )


def map_symbols_to_ensembl(symbols):
    return map_to_ensembl(
        symbols,
        scope="symbol",
        input_column="Input_symbol",
        symbol_column="Official_symbol",
        cleaner=lambda value: str(value) if pd.notna(value) else "",
    )


def map_mim_to_ensembl(mim_values):
    return map_to_ensembl(
        mim_values,
        scope="mim",
        input_column="MIM_number_clean",
        symbol_column="Mapped_GeneSymbol",
        cleaner=clean_mim_number,
    )
```

```python
from IPython.display import display, Markdown
def print_top_terms(enrichment, source, title, n=3):  ## Only dispaly top three terms 
    terms = (
        enrichment.loc[
            enrichment["source"] == source,
            ["name", "p_value", "intersection_size", "query_size"]
        ]
        .copy()
    )
    if terms.empty:
        print(f"\n{title}: no significant terms.")
        return
    terms["GeneRatio"] = (
        terms["intersection_size"] / terms["query_size"]
    )
    terms = (
        terms.sort_values("p_value")
        .head(n)
        .rename(columns={
            "name": "Term",
            "p_value": "Adjusted_p_value",
            "intersection_size": "Gene_count",
        })
    )

    display(Markdown(f"**{title}**"))

    display(
        terms[
            [
                "Term",
                "Gene_count",
                "GeneRatio",
                "Adjusted_p_value",
            ]
        ]
        .style
        .format({
            "GeneRatio": "{:.3f}",
            "Adjusted_p_value": "{:.3e}",
        })
        .hide(axis="index")
    )

print_top_terms(enrichment, "GO:BP", "Top three GO Biological Process terms")
print_top_terms(enrichment, "GO:MF", "Top three GO Molecular Function terms")
print_top_terms(enrichment, "GO:CC", "Top three GO Cellular Component terms")
print_top_terms(enrichment, "WP", "Top three WikiPathways pathways")
```

```python
import pandas as pd
from openpyxl import load_workbook

excel_file = "Supplement file 1v3.xlsx"
workbook = load_workbook(
    excel_file,
    read_only=True,
    data_only=True,
)

print(workbook.sheetnames)

worksheet = workbook["Table S7"]

rows = worksheet.iter_rows(values_only=True)
column_names = next(rows)

pmi = pd.DataFrame(
    rows,
    columns=column_names,
)
workbook.close()


pmi = pmi.dropna(how="all").reset_index(drop=True)

pmi.columns = pmi.columns.astype(str).str.strip()

print(f"Integrated PMI edges: {len(pmi):,}")
display(pmi.head())
```

```python
import re
from collections import defaultdict

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

plot_data = pmi[
    [
        "source_ensembl",
        "target_ensembl",
        "source_database",
    ]
].copy()

plot_data = plot_data.dropna(
    subset=[
        "source_ensembl",
        "target_ensembl",
        "source_database",
    ]
)

plot_data["source_ensembl"] = (
    plot_data["source_ensembl"]
    .astype(str)
    .str.strip()
)

plot_data["target_ensembl"] = (
    plot_data["target_ensembl"]
    .astype(str)
    .str.strip()
)

def make_undirected_edge(row):
    source = row["source_ensembl"]
    target = row["target_ensembl"]

    return tuple(sorted((source, target)))
plot_data["edge"] = plot_data.apply(
    make_undirected_edge,
    axis=1,
)

def parse_databases(value):

    return {
        database.strip()
        for database in re.split(r"[;,|]+", str(value))
        if database.strip()
    }


plot_data["databases"] = (
    plot_data["source_database"]
    .apply(parse_databases)
)

edge_to_databases = defaultdict(set)

for edge, database_set in zip(
    plot_data["edge"],
    plot_data["databases"],
):
    edge_to_databases[edge].update(database_set)


print(
    "Unique undirected interactions:",
    f"{len(edge_to_databases):,}",
)

# All database names
databases = sorted({
    database
    for database_set in edge_to_databases.values()
    for database in database_set
})

print("Databases:", databases)

database_totals = [
    sum(
        database in database_set
        for database_set in edge_to_databases.values()
    )
    for database in databases
]

intersection_counts = defaultdict(int)

for database_set in edge_to_databases.values():
    membership = tuple(
        database in database_set
        for database in databases
    )

    intersection_counts[membership] += 1

sorted_intersections = sorted(
    intersection_counts.items(),
    key=lambda item: (
        -item[1],
        -sum(item[0]),
        item[0],
    ),
)

memberships = [
    membership
    for membership, count in sorted_intersections
]

counts = [
    count
    for membership, count in sorted_intersections
]

if not counts:
    raise ValueError(
        "No database intersections were found."
    )

plt.rcParams.update({
    "font.family": "DejaVu Serif",
    "font.size": 10,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
})

fig = plt.figure(
    figsize=(
        max(12, len(counts) * 0.48),
        max(7, len(databases) * 0.55 + 4),
    )
)

grid = GridSpec(
    2,
    2,
    width_ratios=[2.2, 6],
    height_ratios=[4, 2],
    wspace=0.05,
    hspace=0.05,
)

ax_top = fig.add_subplot(grid[0, 1])
ax_left = fig.add_subplot(grid[1, 0])
ax_matrix = fig.add_subplot(grid[1, 1])

x = np.arange(len(counts))
y = np.arange(len(databases))

main_color = "steelblue"
inactive_color = "lightgrey"

ax_top.bar(
    x,
    counts,
    color=main_color,
    width=0.65,
)

maximum_count = max(counts)

for position, count in zip(x, counts):
    ax_top.text(
        position,
        count + maximum_count * 0.02,
        f"{count:,}",
        ha="center",
        va="bottom",
        fontsize=9,
        rotation=90 if len(counts) > 20 else 0,
    )

ax_top.set_xlim(-0.5, len(counts) - 0.5)
ax_top.set_ylim(
    0,
    maximum_count * 1.15,
)

ax_top.set_ylabel(
    "Number of interactions"
)

ax_top.set_xticks([])
ax_top.grid(
    axis="y",
    alpha=0.25,
)

ax_top.set_axisbelow(True)

ax_top.spines[
    ["top", "right"]
].set_visible(False)

ax_left.barh(
    y,
    database_totals,
    color=main_color,
)

maximum_total = max(database_totals)

for position, count in zip(
    y,
    database_totals,
):
    ax_left.text(
        count + maximum_total * 0.02,
        position,
        f"{count:,}",
        va="center",
        ha="left",
        fontsize=9,
    )

ax_left.set_xlim(
    0,
    maximum_total * 1.25,
)

ax_left.set_ylim(
    len(databases) - 0.5,
    -0.5,
)

ax_left.set_yticks(y)
ax_left.set_yticklabels(databases)

ax_left.set_xlabel(
    "Unique interactions"
)

ax_left.tick_params(
    axis="y",
    length=0,
)

ax_left.grid(
    axis="x",
    alpha=0.20,
)

ax_left.set_axisbelow(True)

ax_left.spines[
    ["top", "right", "left"]
].set_visible(False)

for column, membership in enumerate(memberships):

    active_y = [
        row
        for row, included in enumerate(membership)
        if included
    ]

    for row in y:
        is_active = row in active_y

        ax_matrix.scatter(
            column,
            row,
            s=180,
            color=(
                main_color
                if is_active
                else inactive_color
            ),
            edgecolor=(
                main_color
                if is_active
                else "white"
            ),
            linewidth=0.5,
            zorder=2,
        )

    if len(active_y) > 1:
        ax_matrix.plot(
            [column, column],
            [min(active_y), max(active_y)],
            color=main_color,
            linewidth=3,
            zorder=1,
        )


ax_matrix.set_xlim(
    -0.5,
    len(counts) - 0.5,
)

ax_matrix.set_ylim(
    len(databases) - 0.5,
    -0.5,
)

ax_matrix.set_xticks(x)
ax_matrix.set_xticklabels(
    [""] * len(x)
)

ax_matrix.set_yticks(y)
ax_matrix.set_yticklabels([])

ax_matrix.tick_params(
    length=0
)

for spine in ax_matrix.spines.values():
    spine.set_visible(False)

for row in y:
    if row % 2 == 0:
        ax_matrix.axhspan(
            row - 0.5,
            row + 0.5,
            color="#f4f4f4",
            zorder=0,
        )
fig.suptitle(
    "PMI database overlap",
    fontsize=16,
    fontweight="bold",
    y=0.98,
)

output_file = "pmi_database_overlap.png"

fig.savefig(
    output_file,
    dpi=900,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()

print(f"Saved: {output_file}")
```

```python
# Compare the output of HotNet2
## V3 means PPI, V4 means PMI
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, PathPatch, Rectangle
from matplotlib.path import Path as MplPath
from pathlib import Path
from IPython.display import display, Image

SCHEME_ORDER = ["Original", "Binary", "Scaled", "Random"]
RUN_ORDER = [(v, s) for v in ("v3", "v4") for s in SCHEME_ORDER]

def read_excel_sheet(filename, sheet_name='TableS9'):
    from openpyxl import load_workbook
    workbook = load_workbook(filename, read_only=True, data_only=True)
    candidates = [sheet_name, 'Table S9', 'TableS9']
    selected = next((name for name in candidates if name in workbook.sheetnames), None)
    if selected is None:
        raise ValueError('Could not find TableS9. Sheets: %s' % workbook.sheetnames)
    rows = workbook[selected].iter_rows(values_only=True)
    headers = next(rows)
    frame = pd.DataFrame(rows, columns=headers).dropna(how='all').reset_index(drop=True)
    workbook.close()
    return frame

def clean_table(frame):
    frame = frame.copy()
    frame.columns = frame.columns.astype(str).str.strip()
    required = {'version', 'scheme', 'rank', 'delta', 'module_index',
                'module_size', 'Ensembl_IDs'}
    missing = sorted(required - set(frame.columns))
    if missing:
        raise ValueError('TableS9 is missing columns: %s' % missing)
    for column in ('version', 'scheme', 'Ensembl_IDs'):
        frame[column] = frame[column].astype(str).str.strip()
    for column in ('rank', 'delta', 'module_index', 'module_size'):
        frame[column] = pd.to_numeric(frame[column], errors='raise')
    return frame

def genes(value):
    return {gene.strip() for gene in str(value).split(';') if gene.strip()}

def highest_delta_rows(frame, version, scheme, min_nodes=None):
    subset = frame[(frame['version'] == version) & (frame['scheme'] == scheme)].copy()
    if subset.empty:
        raise ValueError('No rows for %s %s' % (version, scheme))
    highest = subset['delta'].max()
    subset = subset[np.isclose(subset['delta'], highest)]
    if min_nodes is not None:
        subset = subset[subset['module_size'] >= min_nodes]
    return highest, subset

def combined_gene_set(rows):
    return set().union(*(genes(value) for value in rows['Ensembl_IDs'])) if len(rows) else set()

def jaccard(left, right):
    union = left | right
    return len(left & right) / float(len(union)) if union else 1.0

# Load and validate TableS9
table_s9 = clean_table(read_excel_sheet("Supplement file 1v3.xlsx", "TableS9"))
out_dir = Path("tables9_comparison_plots")
out_dir.mkdir(exist_ok=True)
print(table_s9.groupby(["version", "scheme"])["delta"].nunique())
display(table_s9.head())
```

```python
def plot_eight_jaccard(frame, out_dir):
    labels, sets = [], []
    for version, scheme in RUN_ORDER:
        _, rows = highest_delta_rows(frame, version, scheme)
        labels.append('%s %s' % (version, scheme))
        sets.append(combined_gene_set(rows))
    matrix = np.array([[jaccard(a, b) for b in sets] for a in sets])
    fig, ax = plt.subplots(figsize=(10.5, 9))
    image = ax.imshow(matrix, cmap='Blues', vmin=0, vmax=1)
    ax.set_xticks(range(8)); ax.set_yticks(range(8))
    ax.set_xticklabels(labels, rotation=60, ha='right')
    ax.set_yticklabels(labels)
    ax.set_title('Highest-delta module genes: all eight outputs\nGene Jaccard', fontsize=15)
    for row in range(8):
        for column in range(8):
            value = matrix[row, column]
            ax.text(column, row, '%.2f' % value, ha='center', va='center',
                    color='white' if value >= .68 else '#10202c', fontsize=9)
    cbar = fig.colorbar(image, ax=ax, fraction=.046, pad=.04)
    cbar.set_label('Jaccard similarity')
    fig.tight_layout()
    fig.savefig(out_dir / 'eight_gene_jaccard_highest_delta.png', dpi=300,
                bbox_inches='tight', facecolor='white')
    fig.savefig(out_dir / 'eight_gene_jaccard_highest_delta.pdf',
                bbox_inches='tight', facecolor='white')
    plt.close(fig)
    pd.DataFrame(matrix, index=labels, columns=labels).to_csv(
        out_dir / 'eight_gene_jaccard_highest_delta.tsv', sep='\t')

plot_eight_jaccard(table_s9, out_dir)
display(Image(filename=str(out_dir / "eight_gene_jaccard_highest_delta.png")))
```

```python
def v4_delta_modules(frame, min_nodes):
    subset = frame[(frame['version'] == 'v4') & (frame['scheme'] == 'Original')]
    result = {}
    for delta, rows in subset.groupby('delta'):
        rows = rows[rows['module_size'] >= min_nodes].sort_values('module_index')
        result[float(delta)] = [genes(value) for value in rows['Ensembl_IDs']]
    return result

def plot_alluvial(frame, out_dir, min_nodes):
    delta_modules = v4_delta_modules(frame, min_nodes)
    deltas = sorted(delta_modules)
    universe = set().union(*(module for ds in delta_modules.values() for module in ds))
    outside = '__OUTSIDE__'
    node_genes, orders = {}, {}
    for delta in deltas:
        node_genes[delta] = {}
        assigned = set()
        for index, module in enumerate(delta_modules[delta], 1):
            node = 'M%03d_n%d' % (index, len(module))
            node_genes[delta][node] = set(module); assigned.update(module)
        node_genes[delta][outside] = universe - assigned
        orders[delta] = sorted((n for n in node_genes[delta] if n != outside),
                               key=lambda n: (-len(node_genes[delta][n]), n)) + [outside]
    palette = plt.get_cmap('Set3')
    colors = {d: {n: ((.65, .65, .65, 1) if n == outside else
                       palette(i % palette.N)) for i, n in enumerate(orders[d])} for d in deltas}
    gap = max(.7, len(universe) * .0025)
    total_height = len(universe) + max(len(orders[d]) for d in deltas) * gap
    positions = {}
    for delta in deltas:
        y = total_height; positions[delta] = {}
        for node in orders[delta]:
            size = len(node_genes[delta][node])
            if size:
                positions[delta][node] = (y - size, y); y -= size + gap
    fig, ax = plt.subplots(figsize=(17, max(10, .035 * total_height)))
    xs, width = np.arange(len(deltas), dtype=float), .10
    link_rows = []
    for step in range(len(deltas) - 1):
        left, right = deltas[step], deltas[step + 1]
        lo = {n: positions[left][n][0] for n in positions[left]}
        ro = {n: positions[right][n][0] for n in positions[right]}
        for ln in orders[left]:
            for rn in orders[right]:
                if ln not in positions[left] or rn not in positions[right]: continue
                shared = node_genes[left][ln] & node_genes[right][rn]
                if not shared: continue
                amount = len(shared); ly0, ry0 = lo[ln], ro[rn]
                ly1, ry1 = ly0 + amount, ry0 + amount; lo[ln], ro[rn] = ly1, ry1
                x0, x1 = xs[step] + width / 2, xs[step + 1] - width / 2
                c = (x1 - x0) * .48
                verts = [(x0,ly0),(x0+c,ly0),(x1-c,ry0),(x1,ry0),(x1,ry1),
                         (x1-c,ry1),(x0+c,ly1),(x0,ly1),(x0,ly0)]
                codes = [MplPath.MOVETO,MplPath.CURVE4,MplPath.CURVE4,MplPath.CURVE4,
                         MplPath.LINETO,MplPath.CURVE4,MplPath.CURVE4,MplPath.CURVE4,
                         MplPath.CLOSEPOLY]
                color = colors[right][rn] if rn != outside else colors[left][ln]
                ax.add_patch(PathPatch(MplPath(verts,codes), facecolor=color,
                                       edgecolor='none', alpha=.75, zorder=1))
                link_rows.append((left,right,ln,rn,amount,';'.join(sorted(shared))))
    for x, delta in zip(xs, deltas):
        for node in orders[delta]:
            if node not in positions[delta]: continue
            y0, y1 = positions[delta][node]
            ax.add_patch(Rectangle((x-width/2,y0),width,y1-y0,
                                   facecolor=colors[delta][node],edgecolor='#333',lw=.4,zorder=3))
            label = 'Outside >=%d' % min_nodes if node == outside else node
            if y1-y0 >= 2:
                ax.text(x+.07,(y0+y1)/2,label,ha='left',va='center',fontsize=6,zorder=4)
        ax.text(x,total_height+5,'delta = %.7g' % delta,ha='center',fontweight='bold',fontsize=12)
    ax.annotate('decreasing delta / module expansion',
                xy=(xs[0],total_height+15),xytext=(xs[-1],total_height+15),
                arrowprops=dict(arrowstyle='->',lw=1.8,color='#444'),ha='center',fontsize=11)
    ax.set_xlim(-.35,len(deltas)-.65); ax.set_ylim(-2,total_height+22); ax.axis('off')
    ax.set_title('v4 Original module composition across delta\nmodules with at least %d nodes' %
                 min_nodes, fontsize=16, pad=20)
    fig.savefig(out_dir / 'v4_original_delta_alluvial_min5.png', dpi=300,
                bbox_inches='tight', facecolor='white')
    fig.savefig(out_dir / 'v4_original_delta_alluvial_min5.pdf',
                bbox_inches='tight', facecolor='white')
    plt.close(fig)
    pd.DataFrame(link_rows, columns=['left_delta','right_delta','left_module','right_module',
                 'shared_gene_count','Ensembl_IDs']).to_csv(
        out_dir / 'v4_original_delta_alluvial_links_min5.tsv', sep='\t', index=False)

plot_alluvial(table_s9, out_dir, min_nodes=5)
display(Image(filename=str(out_dir / "v4_original_delta_alluvial_min5.png")))
```
